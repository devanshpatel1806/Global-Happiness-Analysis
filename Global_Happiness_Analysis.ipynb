```python

# ==============================================================
# PROJECT: GLOBAL HAPPINESS REPORT ANALYSIS
# RED & WHITE SKILL EDUCATION
# COMPLETE MENU-DRIVEN JUPYTER NOTEBOOK APPLICATION
# ==============================================================

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets

from IPython.display import display, clear_output

sns.set_theme(style="whitegrid")

# ==============================================================
# 2. LOAD DATASET
# ==============================================================

try:
    df = pd.read_csv("happiness.csv")
    print("Dataset loaded successfully!")

except FileNotFoundError:
    raise FileNotFoundError(
        "happiness.csv not found. Keep the CSV file "
        "in the same folder as this notebook."
    )

# Clean column names
df.columns = df.columns.str.strip()

# Handle missing numerical values
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Remove duplicate records
df.drop_duplicates(inplace=True)

# Save cleaned data
df.to_csv("cleaned_happiness.csv", index=False)

print("Data preprocessing completed!")
print("Total Countries:", len(df))
print("Total Columns:", len(df.columns))

# ==============================================================
# 3. DEFINE IMPORTANT COLUMNS
# ==============================================================

COUNTRY = "Country name"
REGION = "Regional indicator"
HAPPINESS = "Ladder score"

GDP = "Log GDP per capita"
SOCIAL = "Social support"
LIFE = "Healthy life expectancy"

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

countries = sorted(df[COUNTRY].dropna().unique().tolist())
regions = sorted(df[REGION].dropna().unique().tolist())

# ==============================================================
# 4. DATA ANALYSIS FUNCTIONS
# ==============================================================

def view_dataset():

    print("FIRST 10 RECORDS")
    display(df.head(10))

    print("\nLAST 10 RECORDS")
    display(df.tail(10))


def dataset_information():

    print("DATASET INFORMATION")

    print("\nTotal Rows:", df.shape[0])
    print("Total Columns:", df.shape[1])
    print("Total Countries:", df[COUNTRY].nunique())
    print("Total Regions:", df[REGION].nunique())

    print("\nCOLUMN INFORMATION")
    info = pd.DataFrame({
        "Column Name": df.columns,
        "Data Type": df.dtypes.astype(str).values,
        "Missing Values": df.isnull().sum().values
    })

    display(info)

    print("\nSTATISTICAL SUMMARY")
    display(df.describe())


def data_cleaning_summary():

    print("DATA CLEANING SUMMARY")

    print("\nMissing Values:")
    display(df.isnull().sum().to_frame("Missing Values"))

    print("Total Missing Values:", df.isnull().sum().sum())

    print("\nDuplicate Records:", df.duplicated().sum())

    print("\nData Types:")
    display(df.dtypes.to_frame("Data Type"))

    print("\nCleaned Dataset:")
    display(df.head())


def statistical_analysis(column):

    data = df[column].dropna()

    print("STATISTICAL ANALYSIS")
    print("Selected Column:", column)

    print("\nMean:", round(data.mean(), 3))
    print("Median:", round(data.median(), 3))
    print("Mode:", round(data.mode().iloc[0], 3))
    print("Minimum:", round(data.min(), 3))
    print("Maximum:", round(data.max(), 3))
    print("Standard Deviation:", round(data.std(), 3))
    print("Variance:", round(data.var(), 3))

    print("\nComplete Summary:")
    display(data.describe().to_frame("Values"))


def country_analysis(country):

    result = df[df[COUNTRY] == country]

    if result.empty:
        print("Country not found!")
        return

    print("COUNTRY ANALYSIS:", country)

    display(result.T)

    print("\nHAPPINESS FACTORS")

    factors = [GDP, SOCIAL, LIFE,
               "Freedom to make life choices",
               "Generosity",
               "Perceptions of corruption"]

    display(result[factors].T)


def country_comparison(country_a, country_b):

    if country_a == country_b:
        print("Please select two different countries.")
        return

    result = df[df[COUNTRY].isin([country_a, country_b])]

    print("COUNTRY COMPARISON")
    display(result.set_index(COUNTRY)[
        [HAPPINESS, GDP, SOCIAL, LIFE]
    ].T)

    result.set_index(COUNTRY)[
        [HAPPINESS, GDP, SOCIAL, LIFE]
    ].plot(kind="bar", figsize=(11, 6))

    plt.title("Country Happiness Factor Comparison")
    plt.ylabel("Score")
    plt.xticks(rotation=0)
    plt.legend(title="Factors")
    plt.tight_layout()
    plt.show()


def happiness_ranking(rank_type, number):

    if rank_type == "Highest Happiness":

        result = df.nlargest(number, HAPPINESS)
        title = "Top Happiest Countries"

    else:

        result = df.nsmallest(number, HAPPINESS)
        title = "Lowest Happiness Countries"

    print(title)

    display(result[
        [COUNTRY, REGION, HAPPINESS]
    ])

    plt.figure(figsize=(11, 6))

    sns.barplot(
        data=result,
        x=HAPPINESS,
        y=COUNTRY
    )

    plt.title(title)
    plt.xlabel("Happiness Score")
    plt.ylabel("Country")
    plt.tight_layout()
    plt.show()


def regional_analysis(selected_region):

    if selected_region == "All Regions":
        result = df.copy()
    else:
        result = df[df[REGION] == selected_region]

    if result.empty:
        print("No data found.")
        return

    print("REGIONAL ANALYSIS")
    display(
        result.groupby(REGION)[HAPPINESS]
        .agg(["count", "mean", "min", "max"])
        .round(3)
    )

    averages = (
        result.groupby(REGION)[HAPPINESS]
        .mean()
        .sort_values(ascending=False)
    )

    averages.plot(
        kind="bar",
        figsize=(11, 5)
    )

    plt.title("Average Happiness by Region")
    plt.ylabel("Average Happiness Score")
    plt.xticks(rotation=40, ha="right")
    plt.tight_layout()
    plt.show()


def factor_analysis(factor):

    print("FACTOR ANALYSIS:", factor)

    display(
        df[[COUNTRY, factor, HAPPINESS]]
        .sort_values(factor, ascending=False)
        .head(15)
    )

    correlation = df[factor].corr(df[HAPPINESS])

    print("\nCorrelation with Happiness:",
          round(correlation, 3))

    plt.figure(figsize=(9, 6))

    sns.regplot(
        data=df,
        x=factor,
        y=HAPPINESS
    )

    plt.title(f"{factor} vs Happiness Score")
    plt.tight_layout()
    plt.show()


def data_visualization(chart, x, y):

    plt.figure(figsize=(11, 6))

    if chart == "Bar Chart":

        plot_data = df.nlargest(15, y)

        sns.barplot(
            data=plot_data,
            x=y,
            y=COUNTRY
        )

    elif chart == "Line Chart":

        plot_data = df.sort_values(x)

        sns.lineplot(
            data=plot_data,
            x=x,
            y=y,
            marker="o"
        )

    elif chart == "Scatter Plot":

        sns.scatterplot(
            data=df,
            x=x,
            y=y,
            hue=REGION
        )

    elif chart == "Histogram":

        sns.histplot(
            data=df,
            x=x,
            kde=True
        )

    elif chart == "Box Plot":

        sns.boxplot(
            data=df,
            x=x
        )

    elif chart == "Pie Chart":

        if x == y:
            plt.close()
            print("Please select different X and Y columns.")
            return

        plot_data = df.nlargest(10, y)

        plt.pie(
            plot_data[y],
            labels=plot_data[COUNTRY],
            autopct="%1.1f%%",
            startangle=90
        )

    plt.title(chart)
    plt.tight_layout()
    plt.show()


def correlation_analysis():

    correlation = df[numerical_columns].corr()

    print("CORRELATION MATRIX")
    display(correlation.round(3))

    plt.figure(figsize=(12, 9))

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.5
    )

    plt.title("Happiness Factors Correlation Heatmap")
    plt.tight_layout()
    plt.show()


def generate_final_report():

    happiest = df.loc[df[HAPPINESS].idxmax()]
    lowest = df.loc[df[HAPPINESS].idxmin()]

    print("=" * 60)
    print("       GLOBAL HAPPINESS ANALYSIS FINAL REPORT")
    print("=" * 60)

    print("\nGENERAL INFORMATION")
    print("Total Countries:", len(df))
    print("Total Regions:", df[REGION].nunique())

    print("\nHAPPINESS RESULTS")
    print("Average Happiness:", round(df[HAPPINESS].mean(), 3))

    print("Happiest Country:", happiest[COUNTRY])
    print("Highest Score:", happiest[HAPPINESS])

    print("Lowest Happiness Country:", lowest[COUNTRY])
    print("Lowest Score:", lowest[HAPPINESS])

    print("\nAVERAGE HAPPINESS FACTORS")

    for factor in [GDP, SOCIAL, LIFE]:

        print(
            factor, ":",
            round(df[factor].mean(), 3)
        )

        print(
            "Correlation with Happiness:",
            round(df[factor].corr(df[HAPPINESS]), 3)
        )

    print("\nTOP 5 HAPPIEST COUNTRIES")

    display(
        df.nlargest(5, HAPPINESS)[
            [COUNTRY, HAPPINESS]
        ]
    )

    print("\nBOTTOM 5 HAPPIEST COUNTRIES")

    display(
        df.nsmallest(5, HAPPINESS)[
            [COUNTRY, HAPPINESS]
        ]
    )

    print("=" * 60)
    print("REPORT GENERATED SUCCESSFULLY")


def export_results():

    df.to_csv("final_happiness_analysis.csv", index=False)

    df[numerical_columns].describe().to_csv(
        "happiness_statistics.csv"
    )

    df[numerical_columns].corr().to_csv(
        "happiness_correlation.csv"
    )

    print("All files exported successfully!")

    print("1. final_happiness_analysis.csv")
    print("2. happiness_statistics.csv")
    print("3. happiness_correlation.csv")


# ==============================================================
# 5. CREATE INTERACTIVE WIDGETS
# ==============================================================

menu = widgets.Dropdown(
    options=[
        ("Select an Option", "0"),
        ("1. View Dataset", "1"),
        ("2. Dataset Information", "2"),
        ("3. Data Cleaning Summary", "3"),
        ("4. Statistical Analysis", "4"),
        ("5. Country Analysis", "5"),
        ("6. Country Comparison", "6"),
        ("7. Happiness Ranking", "7"),
        ("8. Regional Analysis", "8"),
        ("9. GDP Analysis", "9"),
        ("10. Social Support Analysis", "10"),
        ("11. Life Expectancy Analysis", "11"),
        ("12. Data Visualization", "12"),
        ("13. Correlation Analysis", "13"),
        ("14. Generate Final Report", "14"),
        ("15. Export Results", "15"),
        ("16. Exit", "16")
    ],
    description="Menu:",
    layout=widgets.Layout(width="75%")
)

country_widget = widgets.Dropdown(
    options=countries,
    description="Country:",
    layout=widgets.Layout(width="75%")
)

country1_widget = widgets.Dropdown(
    options=countries,
    value=countries[0],
    description="Country 1:",
    layout=widgets.Layout(width="75%")
)

country2_widget = widgets.Dropdown(
    options=countries,
    value=countries[1],
    description="Country 2:",
    layout=widgets.Layout(width="75%")
)

column_widget = widgets.Dropdown(
    options=numerical_columns,
    value=HAPPINESS,
    description="Column:",
    layout=widgets.Layout(width="75%")
)

ranking_widget = widgets.Dropdown(
    options=["Highest Happiness", "Lowest Happiness"],
    description="Ranking:",
    layout=widgets.Layout(width="75%")
)

number_widget = widgets.IntSlider(
    value=10,
    min=5,
    max=30,
    step=1,
    description="Countries:",
    continuous_update=False,
    layout=widgets.Layout(width="75%")
)

region_widget = widgets.Dropdown(
    options=["All Regions"] + regions,
    description="Region:",
    layout=widgets.Layout(width="75%")
)

factor_widget = widgets.Dropdown(
    options=[GDP, SOCIAL, LIFE],
    description="Factor:",
    layout=widgets.Layout(width="75%")
)

chart_widget = widgets.Dropdown(
    options=[
        "Bar Chart",
        "Line Chart",
        "Scatter Plot",
        "Histogram",
        "Box Plot",
        "Pie Chart"
    ],
    description="Chart:",
    layout=widgets.Layout(width="75%")
)

x_widget = widgets.Dropdown(
    options=numerical_columns,
    value=GDP,
    description="X-axis:",
    layout=widgets.Layout(width="75%")
)

y_widget = widgets.Dropdown(
    options=numerical_columns,
    value=HAPPINESS,
    description="Y-axis:",
    layout=widgets.Layout(width="75%")
)

# ==============================================================
# 6. OUTPUT AREA AND RUN BUTTON
# ==============================================================

output = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #cccccc",
        padding="15px",
        width="98%"
    )
)

run_button = widgets.Button(
    description="Run Analysis",
    button_style="success",
    icon="play",
    layout=widgets.Layout(width="200px", height="40px")
)

# ==============================================================
# 7. MAIN MENU FUNCTION
# ==============================================================

def run_analysis(button):

    with output:

        clear_output(wait=True)

        selected = menu.value

        print("GLOBAL HAPPINESS ANALYZER")
        print("-" * 50)

        try:

            if selected == "0":
                print("Please select an option from the menu.")

            elif selected == "1":
                view_dataset()

            elif selected == "2":
                dataset_information()

            elif selected == "3":
                data_cleaning_summary()

            elif selected == "4":
                statistical_analysis(column_widget.value)

            elif selected == "5":
                country_analysis(country_widget.value)

            elif selected == "6":
                country_comparison(
                    country1_widget.value,
                    country2_widget.value
                )

            elif selected == "7":
                happiness_ranking(
                    ranking_widget.value,
                    number_widget.value
                )

            elif selected == "8":
                regional_analysis(region_widget.value)

            elif selected == "9":
                factor_analysis(GDP)

            elif selected == "10":
                factor_analysis(SOCIAL)

            elif selected == "11":
                factor_analysis(LIFE)

            elif selected == "12":
                data_visualization(
                    chart_widget.value,
                    x_widget.value,
                    y_widget.value
                )

            elif selected == "13":
                correlation_analysis()

            elif selected == "14":
                generate_final_report()

            elif selected == "15":
                export_results()

            elif selected == "16":
                print("Thank you for using Global Happiness Analyzer!")
                print("Application closed.")

        except Exception as error:

            print("An error occurred:")
            print(str(error))

# ==============================================================
# 8. CONNECT BUTTON
# ==============================================================

run_button.on_click(run_analysis)

# ==============================================================
# 9. DISPLAY COMPLETE APPLICATION
# ==============================================================

heading = widgets.HTML(
    value="""
    <div style="
        background:linear-gradient(90deg,#17365D,#2874A6);
        color:white;
        padding:20px;
        border-radius:10px;
        text-align:center;">

        <h1>🌍 GLOBAL HAPPINESS ANALYZER</h1>

        <h3>Interactive Data Analysis System</h3>

        <p>World Happiness Report Analysis</p>

    </div>
    """
)

instructions = widgets.HTML(
    value="""
    <h3>📋 Main Menu</h3>
    <p>Select an operation, adjust the required settings,
    and click Run Analysis.</p>
    """
)

application = widgets.VBox([

    heading,

    instructions,

    menu,

    widgets.HTML("<hr><h4>Country Settings</h4>"),

    country_widget,

    country1_widget,

    country2_widget,

    widgets.HTML("<hr><h4>Analysis Settings</h4>"),

    column_widget,

    ranking_widget,

    number_widget,

    region_widget,

    factor_widget,

    widgets.HTML("<hr><h4>Visualization Settings</h4>"),

    chart_widget,

    x_widget,

    y_widget,

    run_button,

    widgets.HTML("<hr><h3>📊 Analysis Results</h3>"),

    output

])

display(application)

print("Global Happiness Analyzer is ready!")
print("Select an option and click Run Analysis.")
```
